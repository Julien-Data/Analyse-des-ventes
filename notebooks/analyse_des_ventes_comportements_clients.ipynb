Ce notebook propose une analyse exploratoire, statistique et business des données commerciales de la librairie Lapage.

L’objectif est double :
- dresser un état des lieux détaillé de l’activité (ventes, clients, produits) ;

Les étapes principales sont les suivantes :

1. **Importation, exploration initiale et nettoyage des données**
2. **Analyse exploratoire** : exploration des 3 datasets, traitement des doublons, valeurs manquantes, types, formats et outliers.

   - répartition par catégorie,

### Gestion des outliers :

```python
#Visualisation des outliers
g = sns.catplot(data = df_data, x = 'price', kind = 'box')
g.fig.suptitle('Distribution des prix et outliers', fontsize=14, fontweight= 'bold' )
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#Création d'un dataframe sans outliers selon le critère de Tukey
Q1 = df_data['price'].quantile(0.25)
Q3 = df_data['price'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - IQR * 1.5
upper = Q3 + IQR * 1.5
df_data_SANS_OUTLIERS = df_data[(df_data['price'] >= lower) & (df_data['price'] <= upper)]
pourcentage = round( len(df_data_SANS_OUTLIERS)/len(df_data) *100, 2)
print("Pourcentage des données de base, après suppression des outliers :" ,pourcentage , "%")
```

En appliquant le critère de Tukey (± 1.5 × IQR), j’obtiens un nouveau DataFrame, df_data_SANS_OUTLIERS, qui conserve environ **93,77 %** des observations initiales.

### Courbe de Lorenz : Analyse de la concentration du chiffre d'affaires :

```python
# Préparation à la Courbe de Lorenz
Ca_par_client = df_data.groupby('client_id')[['price']].sum()
sorted_values = np.sort(Ca_par_client['price'].values)

# Cumul des valeurs
cumulative = np.cumsum(sorted_values)

# Normalisation
cumulative_percent = cumulative / cumulative[-1]  # cumulé / total
x = np.linspace(0.0, 1.0, len(cumulative_percent))

# Ajout de 0 au début pour la courbe
x = np.insert(x, 0, 0)
cumulative_percent = np.insert(cumulative_percent, 0, 0)
```

```python
#Courbe de Lorenz
plt.figure(figsize=(8, 6))
plt.plot(x, cumulative_percent, label='Courbe de Lorenz', color='blue')
plt.plot([0, 1], [0, 1], linestyle='--', color='gray', label='Égalité parfaite')

x_median = np.interp(0.5, cumulative_percent, x) 
plt.plot([0, x_median], [0.5, 0.5], color='red', linestyle=':', label='50% du CA')
plt.plot([x_median, x_median], [0, 0.5], color='red', linestyle=':')

plt.title("Courbe de Lorenz - Répartition du chiffre d'affaires par client", y=1.05, fontweight = 'bold')
plt.xlabel("Proportion de clients")
plt.ylabel("Proportion du chiffre d'affaires")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Indice de GINI
gini = 1 - 2 * np.trapezoid(cumulative_percent, x)
print(f"Indice de Gini : {gini:.4f}")
```

On note des inégalités **modérées à fortes**, une minorité de clients, environ **22 %**,  génèrent **50 %** du chiffre d'affaires.  
Cette distribution suggère un effet de type **Pareto**, indiquant une dépendance significative à une minorité de clients.  
Il serait pertinent d’envisager une **stratégie commerciale différenciée**, ciblant en priorité ces clients à forte contribution (fidélisation, offres personnalisées, suivi renforcé).

### Analyse de cohortes

```python
# Extraction du mois de commande pour regrouper les achats dans le temps
df_cohort = df_data.copy()
df_cohort['order_month'] = df_cohort['date'].dt.to_period('M')

# Date du premier achat
df_cohort['cohort'] = df_cohort.groupby('client_id')['date'].transform('min').dt.to_period('M')

#Index de cohorte
def cohort_index(df):
    return (
        (df['order_month'].dt.year - df['cohort'].dt.year) * 12 +
        (df['order_month'].dt.month - df['cohort'].dt.month))

df_cohort['cohort_index'] = cohort_index(df_cohort)
```

```python
# transforme les données transactionnelles en une matrice de rétention de cohortes
cohort_data = (
    df_cohort.groupby(['cohort', 'cohort_index'])['client_id']
      .nunique()
      .reset_index()
)

cohort_pivot = cohort_data.pivot(
    index='cohort',
    columns='cohort_index',
    values='client_id'
)

cohort_size = cohort_pivot.iloc[:, 0]
retention = cohort_pivot.divide(cohort_size, axis=0)
```

```python
# Matrice de l'analyse de cohorte
purple_cmap = LinearSegmentedColormap.from_list( "custom_purple", ["#f3e9f5", "#c9a3d6", "#8b5fbf", "#5a3a7a"])

plt.figure(figsize=(12, 6))

ax = sns.heatmap(retention, annot=True, fmt=".0%", cmap=purple_cmap, linewidths=0.3,
                 cbar_kws={"label": "Taux de rétention"})

# adaptation automatique de la couleur du texte
for text in ax.texts:
    value = float(text.get_text().strip('%')) / 100

    if value > 0.5:
        text.set_color("white")
    else:
        text.set_color("#2b2b2b")

    text.set_fontsize(8)

plt.title("Analyse de cohorte – Taux de rétention client", pad=20)
plt.ylabel("Cohorte (mois du 1er achat)")
plt.xlabel("Mois depuis le 1er achat")
plt.xticks(rotation=0)
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
# Calcul du taux moyen de rétention par mois depuis la première commande
avg_retention = retention.iloc[:, :6].mean()

print(f"Taux moyen de rétention à M+1 : {avg_retention[1]:.1%}")
print(f"Taux moyen de rétention à M+5 : {avg_retention[5]:.1%}")
```

```python
# Mesure de la dégradation de la rétention entre M+1 et M+5
drop = avg_retention[1] - avg_retention[5]
print(f"Baisse de rétention entre M+1 et M+5 : {drop:.1%}")
print(f"Taux de survie après 5 mois : {avg_retention[5]:.1%}")
```

##  Conclusion – Analyse de cohortes clients

L’analyse des cohortes met en évidence une **chute significative de la rétention dès le premier mois**, avec un taux moyen de réachat de **45,9 % à M+1**.

En revanche, la décroissance est ensuite plus progressive, atteignant **37,7 % à M+5**, soit une baisse limitée de **8,1 points**.

Ces résultats suggèrent que la **principale perte de clients intervient très tôt dans le cycle de vie**, probablement au niveau de l’activation ou du premier réachat. Une fois cette étape franchie, les clients présentent un comportement **relativement stable dans le temps**.

L’enjeu principal pour l’entreprise réside donc dans **l’amélioration de la rétention à court terme**, plutôt que dans la fidélisation long terme.

```python
# Test
contingency_table = pd.crosstab(df_data['categ'], df_data['sex'])
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print(f"Chi² = {chi2}")
print(f"p-value = {p_value}")
```

```python
# Effectif total
n = contingency_table.sum().sum()

# Cramer’s V
cramers_v = np.sqrt(chi2 / (n * (min(contingency_table.shape)-1)))
print(f"Cramer's V = {cramers_v:.4f}")
```

```python
contingency_table
```

```python
# TEST de proportion à deux échantillons
categories = df_test_1['categ'].unique()

for cat in sorted(categories):
    subset = df_test_1[df_test_1['categ'] == cat].set_index('sex')
    
    # Extraire les succès et les nobs pour f et m
    successes = [subset.loc['f', 'compte'], subset.loc['m', 'compte']]

    nobs = [subset.loc['f', 'total'], subset.loc['m', 'total']]
    
    z_score, pval = proportions_ztest(successes, nobs)
    
    print(f"\nCatégorie {cat}:")
    print(f"  - Proportion femmes : {successes[0] / nobs[0]:.2%}")
    print(f"  - Proportion hommes : {successes[1] / nobs[1]:.2%}")
    print(f"  - Z = {z_score:.4f}, p-value = {pval:.4f}")
    
    # Interprétation
    alpha = 0.05
    if pval < alpha:
        print(f"  →  Différence significative entre hommes et femmes (p < {alpha})")
    else:
        print(f"  →  Aucune différence significative (p ≥ {alpha})")
```

Il existe une association **statistiquement significative** entre le sexe des clients et les catégories de livres achetées (Chi², p < 0,05).
L’intensité de cette association est **modérée à forte** (Cramer’s V = 0,46), ce qui indique que le sexe du client a un impact réel sur la catégorie de livres choisie.
